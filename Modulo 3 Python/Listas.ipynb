## Ejercicio básico: lista de útiles escolares
## lápiz, borrador, cuaderno, regla, lápiz

* Agrega un marcador al final de la lista. ✅
* Cuenta cuántas veces aparece lápiz.      ✅
* Busca en qué posición está cuaderno.     ✅
* Inserta sacapuntas en la posición 2.     ✅
* Elimina la primera aparición de lápiz.   ✅
* Elimina el último elemento de la lista.  ✅
* Haz una copia de la lista.               ✅
* Agrega al final estos elementos: pegamento y tijeras. ✅
* Invierte el orden de la lista.           ✅
* Ordena la lista alfabéticamente.         ✅
* Vacía toda la lista.                     ✅

```python
utiles_escolares=["lápiz", "borrador", "cuaderno", "regla", "lápiz"]
print(utiles_escolares)
print(f"Cuantos elementos tiene la lista: {len(utiles_escolares)}") # Imprime la cantidad de elementos en la lista utilizando la función len()
utiles_escolares.append("resaltador") # Agrega "resaltador" al final de la lista
print(utiles_escolares)
print("cuantos lapiz aparecen: " , utiles_escolares.count("lápiz"))# Cuenta cuántas veces aparece "lápiz" en la lista
print("En que posicion esta el cuaderno: " , utiles_escolares.index("cuaderno"))
utiles_escolares.insert(4, "sacapuntas") # Inserta "sacapuntas" en la posición 4 (índice 4)
print("Agregamos elemento nuevo en la pocicion 4:" ,utiles_escolares)
utiles_escolares.remove("lápiz") # Elimina la primera aparición de "lápiz" en la lista
print("Eliminamos el primer elemento de la lista: " ,utiles_escolares)
utiles_escolares.pop() # Elimina el último elemento de la lista y lo devuelve (en este caso, "resaltador")
print("Eliminamos el ultimo elemento: " , utiles_escolares)
copy_lista = utiles_escolares.copy() # Crea una copia de la lista original
print("Creamos una copia de la lista original: " , utiles_escolares)
utiles_escolares.extend(["pegamento" , "tijeras"]) # Agrega los elementos "pegamento" y "tijeras" al final de la lista
print("Se agregaro dos elementos a la lista: " , utiles_escolares)
utiles_escolares.reverse() # Invierte el orden de los elementos en la lista
print("invertimos el orden de la lista: " , utiles_escolares)
utiles_escolares.sort() # Ordena los elementos de la lista en orden alfabético
print("Lista Ordenada alfabeticamente:" , utiles_escolares)
utiles_escolares.clear()  # Elimina todos los elementos de la lista, dejándola vacía
print("Vaciamos la lista: ", utiles_escolares)


```
